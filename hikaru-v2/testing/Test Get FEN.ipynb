```python
import os  # for file management
import chess  # for chess board
import chess.pgn as pgn  # for pgn files
import numpy as np  # for saving the data
import chime
```

```python
# Set the path to the pgn files directory
path = "C:\\Users\\Momad\\OneDrive - Arab Academy for Science and Technology\\Projects\\Project - DRLCE for Chess Robot RL\\CCRL Dataset\\Reformated Dataset"
```

```python
# Add all file names in a directory to a list
num_files = 200000


def add_files(directory, n):
    files = []
    count = 0
    for file in os.listdir(directory):
        if file.endswith(".pgn"):
            files.append(file)
            count += 1
            if count == n:
                break
    return files


pgn_files = add_files(path, num_files)
```

```python
# Add the path to the files
pgn_files = [
    os.path.join(
        path,
        file,
    )
    for file in pgn_files
]
```

```python
# Show the number of files
len(pgn_files)
```

```python
# Read all pgn files in a directory and return a list of games
def read_pgn(files):
    games = []
    for i, file in enumerate(files):
        pgn_file = open(file)
        game = pgn.read_game(pgn_file)
        games.append(game)
        print(f"File {i} read successfully", end="\r")

    return games


pgn_games = read_pgn(pgn_files)
```

```python
# Get the number of games in the dataset
len(pgn_games)
```

```python
# Iterate through the moves of the all the games and add them to a list of moves list
def get_moves(games, n):
    all_moves = []
    for i, game in enumerate(games):
        moves = []
        print(f"Started reading game {i}", end="\r")
        n = np.random.randint(3, 11)

        try:
            num_moves = int(len(list(game.mainline_moves())) / 2)
        except:
            continue

        for j, move in enumerate(game.mainline_moves()):
            if j >= num_moves - n:
                break
            moves.append(move)

        print(f"Game {i} read successfully", end="\r")

        all_moves.append(moves)
    return all_moves


all_games_moves = get_moves(pgn_games, 10)
```

```python
# Get the fen of a game from a list of moves
def get_fen(moves):
    board = chess.Board()
    for move in moves:
        board.push(move)
    return board.fen()


all_games_fen = [get_fen(moves) for moves in all_games_moves]
```

```python
# Remove all the fens that has black to move
all_games_fen = [fen for fen in all_games_fen if fen.split(" ")[1] == "w"]
len(all_games_fen)
```

```python
# Remove all the fens that doesnt have an empty 8th rank
all_games_fen = [fen for fen in all_games_fen if fen.split(" ")[0].split("/")[0] == "8"]
len(all_games_fen)
```
